```python
#Bài 62: Viết chương trình phân tích số nguyên n thành các thừa số nguyên tố trong Python.
def phan_tich(n):
  a = n
  uoc_nguyen_to = []
  i = 2
  while i*i <= n:
    while n%i == 0:
      uoc_nguyen_to.append(i)
      n //= i
    i += 1
  if n > 1:
    uoc_nguyen_to.append(n)
  b = "x".join(map(str, uoc_nguyen_to))
  print(a, '=',b)
n = int(input('Nhập số n: '))
phan_tich(n)
```

```python
#Bài 92:Nhập vào một xâu s từ bàn phím, tạo một xâu mới gồm các chữ số có trong xâu s và in kết quả ra màn hình
a = input('Nhập xâu: ')
b = str(a)
c = len(b)
e = ''
for i in range(0,c):
  d = b[i]
  if '0' <= d <= '9':
     e = e+d
print(e)
```

```python
#Bài 93: Nhập vào một xâu s từ bàn phím. Hãy cho biết xâu s có bao nhiêu kí tự là chữ cái tiếng anh(không phân biệt chữ thương và chữ hoa)
a = input('Nhập xâu: ')
b = str(a)
c = len(b)
e = 0
for i in range(0,c):
  d = b[i]
  if 'a' <= d <= 'z' or 'A' <= d <= 'Z':
     e +=1
print('có: ',e,'kí tự tiếng Anh')
```

```python
#Bài 94: Nhập vào một xâu s từ bàn phím và một kí tự x bất kì. Hãy cho biết kí tự x xuất hiện trong xâu s bao nhiêu lần và thông báo kết quả ra màn hình
a = input('Nhập xâu: ')
k = input('Nhập kí tự cần đếm: ')
b = len(a)
s = 0
for i in range(0,b):
  d = a[i]
  if d == k:
    s += 1
  else:
    s +=0
print('Có',s,'kí tự',k)
```

```python
#Bài 95: Nhập vào một xâu s từ bàn phím, tạo xâu s2 từ xâu s bằng cách loại bỏ các dấu cách có trong xâu s.
a = input('Nhập xâu: ')
b = a.replace(' ','')
print(b)
```

```python
#Bài 96:Nhập xâu họ tên s, hãy chuẩn hóa lại xâu họ tên đó. Xâu họ tên là xâu chuẩn hóa khi thỏa mãn các điều sau:
a = input('Nhập xâu: ')
b = len(a)
a = a[0].upper()+a[1:]
B = []
for i in range(1,b):
  if a[i] == ' ':
    B.append(i+1)
for i in B:
  a = a[:i]+ a[i].upper()+a[i+1:]
print(a)
```
