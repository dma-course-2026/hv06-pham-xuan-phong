```python
#Bài 49: Số Armstrong
def armstrong():
  k = []
  for a in range(1,10):
    for b in range(0,10):
      for c in range(0,10):
        x = a*100+ b *10 + c
        y = a**3 + b**3 + c**3
        if x == y:
          k.append(x)
  print(k)
armstrong()
```
