```python
#Bài 7 :
def tinhtrungbinh(a, b):
    """Hàm này dùng để tính tổng của hai số a và b.
    Tham số:
    a: Số thứ nhất
    b: Số thứ hai
    Trả về:
    Tổng giá trị trung bình của a và b
    """
    return (a + b)/2
print("Tài liệu của hàm abs()")
print(abs.__doc__)

print("Tài liệu của hàm int()")
print(int.__doc__)

print("Tài liệu của hàm input()")
print(input.__doc__)

print("Tài liệu của hàm tự định nghĩa (tinh_tong)")
print(tinhtrungbinh.__doc__)
```

```python
#Bài 11:
import math
C = 50
H = 30
A = []
D = map(int, input('Nhập giá trị D: ').split(','))
for d in D:
    if d < 0:
        A.append('Không tính được')
    else:
        Q = ((2*C*d)/H)**0.5
        A.append(str(Q))
print(','.join(A))
```

```python
#Bài 16:
B = []
def nhiphan(n):
  a = 0
  for i in n:
    a = a*2+int(i)
  return(a)
A = input('Nhập các chuỗi số nhị phân').split(',')
for i in A:
  b = nhiphan(i)
  if b%5==0:
    B.append(i)
print(','.join(B))
```

```python
#Bài 18:
n = input('Nhập 1 câu: ')
chucai = 0
chuso = 0
for i in n:
  if 'a'<=i<='z' or 'A'<=i<='Z':
    chucai+=1
  elif '0'<=i<='9':
    chuso+=1
print('Số chữ cái là: ',chucai)
print('Số chữ số là: ', chuso)
```

```python
#Bài 22:
so_tien = 0
while True:
  a = input('')
  if a == '':
    break
  loai,tien = a.split('=')
  if loai == 'D':
    so_tien += int(tien)
  elif loai == 'W':
    so_tien -= int(tien)
print('Số tiền: ',so_tien)
```

```python
#Bài 23:
def kiemtradoixung(A):
  B=list(reversed(A))
  if A==B:
    return True
  elif A!=B:
    return False
A = input('Nhập một dãy số: ').split(',')
if kiemtradoixung(A):
  print('Dãy đó đối xứng')
else:
  print("Dãy đó không đối xứng")
```
