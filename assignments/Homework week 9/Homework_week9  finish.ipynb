```python
#Bài 26:
class VietNam():
 def __init__(self,nhandan,datnuoc):
  self.nhandan=nhandan
  self.datnuoc=datnuoc
  def moinguoi(self):
    print(f"{self.nhandan} thân thiện")
  def nuoc(self):
    print(f"{self.datnuoc} phát triển")
class Hanoi(VietNam):
  def nguoihanoi(self):
    print(f"{self.nhandan} tài năng,giàu có")
hn = Hanoi('Người Hà Nội','Người Việt Nam')
hn.nguoihanoi()
print(hn.nhandan)
print(hn.datnuoc)
```

```python
#Bài 27:
class Circle():
  def __init__(self,r):
    self.r = r
  def dientich(self):
    return(3.14*self.r**2)
c = Circle(4)
print(c.dientich())
```

```python
#Bài 28:
class Hinhchunhat():
  def __init__(self,chieudai,chieurong):
    self.chieudai=chieudai
    self.chieurong=chieurong
  def dientich(self):
    return(self.chieudai*self.chieurong)
a = Hinhchunhat(4,5)
print(a.dientich())
```
